```python
# function to open zarr store with a provided s3 bucket path
def open_zarr_store(s3_path, s3_options):
    # initalize s3 file system
    s3 = s3fs.S3FileSystem(**s3_options)

    # define location of zarr store and open
    store = s3fs.S3Map(root=s3_path, s3=s3, check=False)
    zarr_store = xr.open_zarr(store)
    
    return zarr_store
```

## Add observations

Data are downloaded from the Arctic Great Rivers Observatory here: https://arcticgreatrivers.org/discharge/

Main website here: https://arcticgreatrivers.org/

```python
# Read the Excel spreadsheet
# replace "discharge_data.xlsx" with your filename
df = pd.read_excel("~/efs-mount-point/mzahn/sassie/Mack_runoff/Mackenzie_ArcticRedRiver_Version_20250327.xlsx")

# Ensure date is datetime
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# Convert discharge to numeric (turns "NA" into NaN automatically)
df['discharge'] = pd.to_numeric(df['discharge'], errors='coerce')
```

```python
# Create an xarray DataArray
discharge_da = xr.DataArray(
    df['discharge'].values,
    coords={"time": df['date'].values},
    dims=["time"],
    name="discharge"
)
```

```python
# Plot using xarray’s built-in plot (matplotlib under the hood)
discharge_da.plot.line(figsize=(12, 2))

plt.title("Mackenzie River Discharge")
plt.xlabel("Date")
plt.ylabel("Discharge (m³/s)")
plt.grid(True)
plt.show()
```

```python
# Resample to monthly mean
monthly_da = discharge_da.resample(time='M').mean()

# Group by year
monthly_by_year = monthly_da.groupby('time.year')

plt.figure(figsize=(8,3))

# Loop through each year and plot the monthly time series
for year, da_year in monthly_by_year:
    # Use the month number for x-axis (1-12)
    da_year_months = da_year['time.month']
    plt.plot(da_year_months, da_year.values, label=str(year))

plt.xlabel('Month')
plt.ylabel('Monthly Mean Discharge')
plt.title('Monthly Mean Discharge by Year')
plt.xticks(range(1,13))
# plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', ncol=1, fontsize='small')
plt.tight_layout()
plt.show()
```

```python
# Calculate daily climatology (group by DOY)
obs_clim_doy = discharge_da.sel(time=slice("2014","2020")).groupby('time.dayofyear').mean('time')
# obs_clim_doy = discharge_da.groupby('time.dayofyear').mean('time')

# Create time coordinate for 2020 (DOY → actual dates)
time_coord = pd.date_range(start='2020-01-01', end='2020-12-31', freq='D')

# Assign the time coordinate
obs_clim_da = obs_clim_doy.copy()
obs_clim_da = obs_clim_da.assign_coords(dayofyear=time_coord)

# Optional: drop the old 'dayofyear' coordinate to avoid confusion
obs_clim_da = obs_clim_da.rename({'dayofyear':'time'})
```

```python
# model climatology
model_clim = (runoff_da_beaufort.sel(time='2020')*beaufort_area).sum(dim=['i','j'])
```
